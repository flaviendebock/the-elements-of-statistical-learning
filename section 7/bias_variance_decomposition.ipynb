# Bias-Variance Decomposition

Consider the model $Y = f\left(\boldsymbol{x}\right)+\varepsilon$, where $\varepsilon$ is independent of the training data and $\mathbb{E}\varepsilon =0, \mathbb{V}\varepsilon = \sigma^2$. (Implying that $\mathbb{E}Y = f\left(\boldsymbol{x}\right)$). 

For a regression fit $\hat{f}\left(\boldsymbol{x}\right)$, the expected prediction error at a point $\boldsymbol{x=x_0}$ is:

\begin{align*}
Err\left(\boldsymbol{x_0}\right)&=\mathbb{E}\left[\left(Y-\hat{f}\left(\boldsymbol{x}\right)\right)^2|\boldsymbol{x=x_0}\right] \mathbb{E}\left[\left(f\left(\boldsymbol{x_0}\right)+\varepsilon -\hat{f}\left(\boldsymbol{x_0}\right)\right)^2\right]\\
&= \mathbb{E}\varepsilon^2 + \mathbb{E}\left[\left(f\left(\boldsymbol{x_0}\right)-\hat{f}\left(\boldsymbol{x_0}\right)\right)^2\right] + \mathbb{E}\left[\epsilon\left(f\left(\boldsymbol{x_0}\right)-\hat{f}\left(\boldsymbol{x_0}\right)\right)\right] \\
&= \sigma^2 + \mathbb{E}\left[\left(f\left(\boldsymbol{x_0}\right)-\mathbb{E}\left(\hat{f}\left(\boldsymbol{x_0}\right)\right)+\mathbb{E}\left(\hat{f}\left(\boldsymbol{x_0}\right)\right)-\hat{f}\left(\boldsymbol{x_0}\right)\right)^2\right] \\
&= \sigma^2 + \left[f\left(\boldsymbol{x_0}\right)-\mathbb{E}\left(\hat{f}\left(\boldsymbol{x_0}\right)\right)\right]^2+ \mathbb{E}\left[\left(\hat{f}\left(\boldsymbol{x_0}\right)-\mathbb{E}\left(\hat{f}\left(\boldsymbol{x_0}\right)\right)\right)^2\right] \\
&= \sigma^2 + \text{Bias}^2\left(\hat{f}\left(\boldsymbol{x_0}\right)\right)+ \mathbb{V}\left(\hat{f}\left(\boldsymbol{x_0}\right)\right).
\end{align*}

Typically, the more complex we make the model $\hat{f}$, the lower the bias but the higher the variance. 


For instance, for a linear fit by the least squares method, we have:

$$\hat{f}\left(\boldsymbol{x}\right)=\boldsymbol{x'\hat{\beta}= x'\left(X'X\right)^{-1}X'y},$$

where $\boldsymbol{x}\in\mathbb{R}^p, \boldsymbol{X}\in\mathbb{R}^{N\times p}$ is the design matrix, $\boldsymbol{\hat{\beta}}\in\mathbb{R}^p, \boldsymbol{y}\in\mathbb{R}^N$. Define:

$$\boldsymbol{h\left(x_0\right)=X\left(X'X\right)^{-1}x_0},$$

then $\boldsymbol{h\left(x_0\right)'=x_0'\left(X'X\right)^{-1}X'}$, so that $\hat{f}\left(\boldsymbol{x_0}\right)=\boldsymbol{h\left(x_0\right)'y = h\left(x_0\right)'f\left(x_0\right)+h\left(x_0\right)'\varepsilon}$. So that the variance of the fit is:

\begin{align*}
\mathbb{V}\left(\hat{f}\left(\boldsymbol{x_0}\right)\right) &= \mathbb{V}\left(\boldsymbol{h\left(x_0\right)'\varepsilon}\right)= \boldsymbol{h\left(x_0\right)'}\mathbb{V}\left(\epsilon\right)h\left(x_0\right)= \sigma^2||h\left(\boldsymbol{x_0}\right)||^2,
\end{align*}

so that the expected prediction error at a point $\boldsymbol{x=x_0}$ is

$$Err\left(\boldsymbol{x_0}\right) = \sigma^2 + \left[f\left(\boldsymbol{x_0}\right)-\mathbb{E}\left(\hat{f}\left(\boldsymbol{x_0}\right)\right)\right]^2+\sigma^2||h\left(\boldsymbol{x_0}\right)||^2$$

With associated in-sample error:

$$\frac{1}{N}\sum_{i=1}^N\text{Err}\left(\boldsymbol{x}_i\right)=\sigma^2+\frac{1}{N}\sum_{i=1}^N\left[f\left(\boldsymbol{x_i}\right)-\mathbb{E}\left(\hat{f}\left(\boldsymbol{x_i}\right)\right)\right]^2+\sigma^2\frac{1}{N}\sum_{i=1}^N||h\left(\boldsymbol{x_i}\right)||^2$$

From the definition of $h\left(\boldsymbol{x}_i\right)$, construct the matrix:

$$\boldsymbol{H}=\begin{bmatrix}
h\left(\boldsymbol{x_1}\right)|&\cdots&|h\left(\boldsymbol{x_N}\right) 
\end{bmatrix}=\boldsymbol{X\left(X'X\right)^{-1}X'}.$$

Clearly, $\boldsymbol{H}$ is the orthogonal projection matrix onto the space spanned by the \boldsymbol{X}'s (the hat matrix). Clearly, it is idempotent and symmetric. Because of this property, its eigenvalues are either 0 or 1. Indeed, consider an eigenvector $\boldsymbol{v\neq 0}$ with associated eigenvalue $\lambda$. By definition, we obtain that:

\begin{align*}
\boldsymbol{Hv} &= \lambda\boldsymbol{v} \\
\implies \boldsymbol{Hv} &= \lambda\boldsymbol{Hv} \ \text{as } \boldsymbol{H^2 = H} \\
\implies \lambda\boldsymbol{v} &= \lambda^2\boldsymbol{v} \\ 
\implies \lambda &= \lambda^2 \ \text{as } \boldsymbol{v\neq 0} \implies \lambda = 1 \ \text{or } \lambda = 0
\end{align*}

Considering $\boldsymbol{X}$ to be full rank, we get $\mathrm{rank}\left(\boldsymbol{X}\right) = p = \mathrm{rank}\left(\boldsymbol{H}\right)$. Therefore, $\boldsymbol{H}$ has $p$ eigenvalues equal to 1. 

Now, by definition of the Frobenius norm:

$$
\frac{1}{N}\sum_{i=1}^N||h\left(\boldsymbol{x_i}\right)||^2 = ||\boldsymbol{H}||^2_F = \frac{1}{N}\mathrm{tr}\left(\boldsymbol{H'H}\right)=\frac{1}{N}\mathrm{tr}\left(\boldsymbol{HH}\right)=\frac{1}{N}\mathrm{tr}\left(\boldsymbol{H}\right)=\frac{p}{N}.
$$

We finally obtain:


$$\frac{1}{N}\sum_{i=1}^N\text{Err}\left(\boldsymbol{x}_i\right)=\sigma^2+\frac{1}{N}\sum_{i=1}^N\left[f\left(\boldsymbol{x_i}\right)-\mathbb{E}\left(\hat{f}\left(\boldsymbol{x_i}\right)\right)\right]^2+\sigma^2\frac{p}{N}.$$

Here model complexity is directly related to the num-
ber of parameters $p$.

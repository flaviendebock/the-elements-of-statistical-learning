# Forward-Stepwise Selection

Rather than search through all possible subsets (which becomes infeasible
for $p$ much larger than 40), we can seek a good path through them. Forward-stepwise selection starts with the intercept, and then sequentially adds into
the model the predictor that most improves the fit

Consider the linear model

$$\boldsymbol{y = X\beta+\varepsilon},$$

where $\boldsymbol{y}\in\mathbb{R}^N$, $\boldsymbol{X}\in\mathbb{R}^{N\times p}$ is our design matrix, $\boldsymbol{\beta}\in\mathbb{R}^p$ is the associated vector of parameters, and $\boldsymbol{\varepsilon}\in\mathbb{R}^N$ are the error terms. 

Furthermore, consider the QR decomposition of the design matrix:

$$\boldsymbol{X= QR},$$

where $\boldsymbol{Q}\in\mathbb{R}^{N\times p}$ is an orthogonal matrix (i.e $\boldsymbol{Q'Q=I_p}$), and $\boldsymbol{R}\in\mathbb{R}^{p\times p}$ is an upper-triangular matrix. 

The Least Squares problem associated with our linear model can be written as

$$\min_{\boldsymbol{\beta}}||\boldsymbol{y-X\beta}||^2.$$

The associated normal equations are:

$$\boldsymbol{X'X\hat{\beta}=X'y}.$$

Using the QR decomposition and the normal equations we obtain the following expression for $\boldsymbol{\hat{\beta}}:$

$$\boldsymbol{\hat{\beta}=R^{-1}Q'y},$$

and for the fitted values $\boldsymbol{\hat{y}}$:

$$\boldsymbol{\hat{y} = QQ'y}$$

We will use that decomposition later on. Before looking at forward-stepwise selection, consider adding a new feature to our linear model:

$$\boldsymbol{y = X\beta}+\gamma\boldsymbol{z}+\boldsymbol{\varepsilon},$$

where $\boldsymbol{z}\in\mathbb{R}^N$, and $\gamma\in\mathbb{R}$. The key point here is that only the part of $\boldsymbol{z}$ that is not already explained by $\boldsymbol{X}$ can improve the fit, denote it $\boldsymbol{z_{\perp}=M_Xz}$. Where $\boldsymbol{M_X}$ is the orthogonal projection matrix onto the space orthogonal to the space spanned by the $\boldsymbol{X}'s$. 

The residuals from the original model can be written as $\boldsymbol{\hat{\varepsilon} = y-\hat{y}=M_Xy}$. 

By the Frisch-Waugh-Lowell theorem, the estimated coefficient associated with the new feature writes:

$$\hat{\gamma}\boldsymbol{ = \left(z_{\perp}'z_{\perp}\right)^{-1}z_{\perp}'\hat{\varepsilon}}$$

As the improvement in fitted values is given by $\gamma \boldsymbol{z_{\perp}}$, the new residuals are $\boldsymbol{\hat{\varepsilon}- \gamma \boldsymbol{z_{\perp}}}$. Denoting by $RSS_{\boldsymbol{X}}$ the residual sum of squares in the original model and by $RSS_{\boldsymbol{X,z}}$ the residual sum of squares in the model with the new feature, we obtain after some algebra:

$$RSS_{\boldsymbol{X,z}}=RSS_{\boldsymbol{X}}-\frac{\left(\boldsymbol{\hat{\varepsilon}'z_{\perp}}\right)^2}{\boldsymbol{z_{\perp}'z_{\perp}}}.$$

So that the reduction in RSS is:

$$\Delta RSS = \frac{\left(\boldsymbol{\hat{\varepsilon}'z_{\perp}}\right)^2}{\boldsymbol{z_{\perp}'z_{\perp}}}.$$

After some algebra, and using the fact that the orthogonal projection matrix onto the space spanned by the $\boldsymbol{X}'s$, $\boldsymbol{P_X}$, is equal to:

$$\boldsymbol{P_X = X\left(X'X\right)^{-1}X'=QR\left(R'R\right)^{-1}R'Q'= QQ'},$$

as $\boldsymbol{R}$ is invertible as long as there are no zero's on its diagonal (remember that it is an upper-triangular matrix). We get that

$$\Delta RSS = \boldsymbol{\frac{\left(z_{\perp}'\hat{\varepsilon}\right)^2}{||z_{\perp}||^2-||Q'z_{\perp}||^2}}$$

The point being that the new feature candidate will be the one with the largest $\Delta RSS$, i.e the one reducing the most the residual sum of squares. Computing this scalar for each feature is less expansive than fitting a linear regression for each of them.
